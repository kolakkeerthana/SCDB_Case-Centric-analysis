To calculate voting or to analyse voting patterns, we cannot proceed with it as a variable for determining unanimous/split decisions. We would be creating a fesute which is vote_margin caluclated using maj and min votes and derive a percentage

```python
case_analysis["vote_margin"] = case_analysis["majVotes"] - case_analysis["minVotes"]
```

```python
print(case_analysis["vote_margin"])
```

```python
case_analysis["majority_vote_pct"] = (
    case_analysis["majVotes"] /
    (case_analysis["majVotes"] + case_analysis["minVotes"])
) * 100
```

Performing EDA

How has Supreme Court case volume changed over time?

```python
cases_by_year = (
    SupremeCourtCaseAnalysis
    .groupby("decision_year")
    .size()
    .reset_index(name="case_count")
)

cases_by_year.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(
    cases_by_year["decision_year"],
    cases_by_year["case_count"]
)

plt.xlabel("Decision Year")
plt.ylabel("Number of Cases")
plt.title("Supreme Court Cases Decided by Year")

plt.show()
```

Processing time

How long does it take for cases to move from oral argument to decision?

```python
processing_summary = (
    SupremeCourtCaseAnalysis["processing_days"]
    .describe()
)

processing_summary
```

```python
processing_by_year = (
    SupremeCourtCaseAnalysis
    .groupby("decision_year")["processing_days"]
    .agg(
        ["count", "mean", "median", "min", "max"]
    )
    .reset_index()
)

processing_by_year.head()
```

Processing time by issue

```python
processing_by_issue = (
    SupremeCourtCaseAnalysis
    .groupby("issueArea_name")["processing_days"]
    .agg(
        ["count", "mean", "median"]
    )
    .sort_values("median", ascending=False)
    .reset_index()
)

processing_by_issue
```

Votinganalysis

```python
SupremeCourtCaseAnalysis[
    ["majVotes", "minVotes", "vote_margin", "majority_vote_pct"]
].describe()
```

```python
voting_by_issue = (
    SupremeCourtCaseAnalysis
    .groupby("issueArea_name")
    .agg(
        cases=("caseId", "count"),
        avg_majority_votes=("majVotes", "mean"),
        avg_minority_votes=("minVotes", "mean"),
        avg_vote_margin=("vote_margin", "mean"),
        avg_majority_pct=("majority_vote_pct", "mean")
    )
    .sort_values("avg_vote_margin")
    .reset_index()
)

voting_by_issue
```

petitioner outcomes:

```python
petitioner_outcomes = (
    SupremeCourtCaseAnalysis["partyWinning_name"]
    .value_counts(dropna=False)
    .reset_index()
)

petitioner_outcomes.columns = [
    "outcome",
    "case_count"
]

petitioner_outcomes
```

by issue area

```python
win_rate_by_issue = (
    SupremeCourtCaseAnalysis
    .groupby("issueArea_name")
    .agg(
        total_cases=("caseId", "count"),
        petitioner_wins=(
            "partyWinning_name",
            lambda x: (x == "Petitioner Won").sum()
        )
    )
)

win_rate_by_issue["petitioner_win_rate"] = (
    win_rate_by_issue["petitioner_wins"]
    / win_rate_by_issue["total_cases"]
    * 100
)

win_rate_by_issue.sort_values(
    "petitioner_win_rate",
    ascending=False
)
```
